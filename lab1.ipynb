## Просмотр информации о наборе данных и предобработка

Метод info() прмиеняемый к объекту класса DataFrame показывает следующую информацию: 
- Сколько есть записей в таблице
- Какие колонки включены в таблицу
- Какой тип данных у записей в колонках
- Сколько информации занимает данная таблица

```python
df.info()
```

Также есть команда для просмотра статистических данных

```python
df.describe()
```

Метод duplicate показывает дублирующиеся данные в наборе данных, sum прмиеняется для подсчёта этих дубликатов

```python
print(df.duplicated().sum())
```

В наборе данных почти 10% дубликатов, от них необходимо избавиться

Для этого будет применены методы drop_duplicates для удаления дубликатов и reset_index для того, чтобы пересчитать индексы в таблице

```python
df = df.drop_duplicates()
df = df.reset_index(drop=True)
df
```

Для просмотра первых строк таблицы используется метод head(n), где n - число строк

```python
df.head(10)
```

Для просмотра последних строк таблицы используется метод tail(n), где n - число строк

```python
df.tail(10)
```

Для доступа к группе строк или столбцов по их обозначениям можно использовать атрибут loc

```python
print(df.loc[0,'Year'])
```

## Манипуляции с DataFrame

Для сортировки значений в DataFrame можно использовать метод .sort_values(by), в котором можно указать основание сортировки, порядок, игнорирование индексов и прочее

```python
df.sort_values(by = 'Price(euro)', ascending=True).head(10)
```

Сохранение DataFrame как отдельного csv файла возможн ос помощью метода to_csv()

```python
df.to_csv('cars_nodup.csv', index=False)
```
